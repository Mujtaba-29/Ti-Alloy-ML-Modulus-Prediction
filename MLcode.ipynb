```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from xgboost import XGBRegressor
from sklearn.neighbors import KNeighborsRegressor
from tabulate import tabulate

# Split data
X = dt.drop(columns=["YM (GPa) "])
y = dt["YM (GPa) "]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Models with best tuned parameters
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0, max_iter=1000),
    "Lasso Regression": Lasso(alpha=0.1, max_iter=1000),
    "ElasticNet Regression": ElasticNet(alpha=0.1, l1_ratio=0.5, max_iter=1000),

    "Random Forest": RandomForestRegressor(
        n_estimators=500,
        min_samples_split=2,
        min_samples_leaf=1,
        max_features='sqrt',
        max_depth=20,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        subsample=0.8,
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        random_state=42
    ),

    "XGBoost": XGBRegressor(
        subsample=0.8,
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        colsample_bytree=0.7,
        random_state=42
    ),

    "K-Nearest Neighbors": KNeighborsRegressor(
        weights='distance',
        p=1,
        n_neighbors=5
    )
}

# Train and evaluate
results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    results[name] = {
        "MSE": mean_squared_error(y_test, pred),
        "R2": r2_score(y_test, pred)
    }

# Display results
print(tabulate(
    [[k, v["MSE"], v["R2"]] for k, v in results.items()],
    headers=["Model", "MSE", "R²"],
    tablefmt="grid"
))
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, r2_score
from tabulate import tabulate

results_table = []

for name, model in models.items():
    y_train_pred = model.predict(X_train)
    y_test_pred  = model.predict(X_test)

    test_r2 = r2_score(y_test, y_test_pred)

    results_table.append([
        name,
        f"{mean_absolute_error(y_train, y_train_pred):.2f}",
        f"{mean_absolute_error(y_test,  y_test_pred):.2f}",
        f"{r2_score(y_train, y_train_pred):.4f}",
        f"{round(test_r2, 2):.2f}",
        test_r2
    ])

df_results = pd.DataFrame(results_table, columns=[
    "Model", "Train MAE (GPa)", "Test MAE (GPa)", "Train R²", "Test R²", "Test R² (sort)"
])

df_results = df_results.sort_values(by="Test R² (sort)", ascending=False).drop(columns=["Test R² (sort)"]).reset_index(drop=True)

print(tabulate(df_results.values, headers=df_results.columns, tablefmt="grid"))
```

```python
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
import numpy as np

# Split data
X = dt.drop(columns=["YM (GPa) "])
y = dt["YM (GPa) "]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Base models with tuned parameters
base_models = [
    ("Gradient Boosting", GradientBoostingRegressor(
        subsample=0.8,
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        random_state=42
    )),
    ("XGBoost", XGBRegressor(
        subsample=0.8,
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        colsample_bytree=0.7,
        random_state=42
    )),
    ("Random Forest", RandomForestRegressor(
        n_estimators=500,
        min_samples_split=2,
        min_samples_leaf=1,
        max_features='sqrt',
        max_depth=20,
        random_state=42
    ))
]

# Stacking Regressor
stack_model = StackingRegressor(
    estimators=base_models,
    final_estimator=LinearRegression(),  # meta-model
    n_jobs=-1
)

# Train stacking model
stack_model.fit(X_train, y_train)

# Predictions & evaluation
y_pred = stack_model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Stacking MSE: {mse:.3f}")
print(f"Stacking RMSE: {np.sqrt(mse):.3f}")
print(f"Stacking R²: {r2:.6f}")
```
